# ChurnLab — Exploratory Data Analysis
IBM Telco Customer Churn (7,043 customers, 21 features).

Questions answered: class balance, missing values, churn by category, churn by tenure, and numeric summaries.

```python
import pandas as pd
from churnlab import config
from churnlab.data import load_raw, clean, engineer, encode

df = load_raw(config.RAW_CSV)
cleaned = clean(df)
eng = engineer(cleaned)
X, y, encoders = encode(eng)
print(f'raw shape: {df.shape}')
print(f'features: {X.shape[1]}')
```

## Class balance
The target is imbalanced — churn is the minority class.

```python
from churnlab.eda import class_balance
class_balance(eng)
```

## Missing values
`TotalCharges` has blanks (short-tenure customers) that we coerce to 0 after cleaning.

```python
missing = df.isna().sum()
missing[missing > 0]
```

## Churn rate by category
Month-to-month contracts and no technical support churn far more.

```python
from churnlab.eda import churn_by_feature
churn_by_feature(eng, 'Contract').round(4)
```

```python
churn_by_feature(eng, 'TechSupport').round(4)
```

## Churn rate by tenure bucket
Shorter tenure churns more.

```python
from churnlab.eda import churn_by_tenure_bucket
churn_by_tenure_bucket(eng).round(4)
```

## Numeric feature summary

```python
from churnlab.eda import feature_summary
feature_summary(eng).round(3)
```

## Takeaways
- **Class imbalance:** ~26.5% churn — accuracy alone is misleading; we report precision/recall/F1 and ROC-AUC.
- **Strongest signal:** `Contract` (month-to-month) and `TechSupport` (No) — both show much higher churn rates.
- **Tenure:** shorter-tenure customers churn more.
- **Derived features:** `tenure_years`, `charges_per_tenure_month`, `num_services`, `has_internet` add signal.
- **Missing data:** `TotalCharges` blanks are coerced to 0; no other columns are missing.
